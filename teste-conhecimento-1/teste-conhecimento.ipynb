##2. Identificação das Variavéis

##Variáveis:
- **mes:** Variável temporal, incremental que vai aumentando um valor a cada linha, representando cada mês.
- **custo_transporte:** Variável dependente que representa o custo de transporte da empresa no mês correspondente.

a) **Variável independente:** O mês, pois é a variável temporal que avança independentemente.

b) **Variável dependente:** O custo de transporte, pois o seu valor varia em função do mês analisado.

c) **Contexto:** O mês representa a linha do tempo (1 a 12 meses observados), enquanto o custo de transporte representa a despesa total (em R$) incorrida pela empresa com frete/logística em cada período.

##3. Gráfico de Dispersão

```python
plt.figure(figsize=(9, 5))
plt.scatter(df['mes'], df['custo_transporte'], color='royalblue', s=60, label='Dados Observados')
plt.title('Custo Mensal de Transporte ao Longo do Tempo', fontsize=13)
plt.xlabel('Mês (x)', fontsize=11)
plt.ylabel('Custo de Transporte em R$ (y)', fontsize=11)
plt.xticks(range(1, 13))
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()
```

##4. Encontrar uma função

```python
# Ajuste linear simples de 1º grau (grau 1 = afim)
a, b = np.polyfit(df['mes'], df['custo_transporte'], 1)

print(f"Coeficiente angular (a): {a:.2f}")
print(f"Coeficiente linear (b): {b:.2f}")
print(f"Função afim encontrada: f(x) = {a:.2f}x + {b:.2f}")
```

##5. Interpretar a Função

a) **Coeficiente $a$:** Representa a taxa de variação média mensal, isto é, o acréscimo médio de custo a cada mês decorrido (~R$ 1.791,61/mês).

b) **Coeficiente $b$:** É o custo estimado no "mês zero" (ponto de partida teórico da reta no eixo $y$, R$ 42.204,55).

c) **Crescimento ou redução:** Como $a > 0$, a reta tem inclinação positiva, indicando crescimento contínuo dos custos.

##6. Representação da função graficamente

```python
x_reta = np.array(range(1, 13))
y_reta = a * x_reta + b

plt.figure(figsize=(9, 5))
plt.scatter(df['mes'], df['custo_transporte'], color='royalblue', s=60, label='Dados Reais')
plt.plot(x_reta, y_reta, color='crimson', linewidth=2, label=f'Tendência: f(x) = {a:.2f}x + {b:.2f}')
plt.title('Ajuste Linear dos Custos de Transporte', fontsize=13)
plt.xlabel('Mês (x)', fontsize=11)
plt.ylabel('Custo de Transporte em R$ (y)', fontsize=11)
plt.xticks(range(1, 13))
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()
```

##7. Previsões

```python
meses_futuros = [13, 14, 15, 18, 24, 36, 48, 96]
previsoes = {m: (a * m + b) for m in meses_futuros}

for m, val in previsoes.items():
    print(f"Custo estimado para o mês {m}: R$ {val:,.2f}")
```

##8. Análise das Previsões

a) **Aderência:** A função afim representa muito bem o comportamento dos custos, pois os pontos reais orbitam rente à reta estimada.

b) **Tendência:** Os custos aumentam consistentemente e constantemente ao longo de todo o histórico.

c) **Aumento mensal:** Os custos crescem aproximadamente $R\$ 1.791,61$ a cada novo mês e aproximadamente $R\$ 42.998,6$ a cada ano.
